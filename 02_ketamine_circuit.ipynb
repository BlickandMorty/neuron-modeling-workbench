## A first calculation you can verify
The excitatory-cell passive time constant is $\tau=C/g_L$. This is a passive-membrane identity, not a fitted brain measurement.

```python
import brian2 as b
tau = (500*b.pF)/(25*b.nS)
print('Excitatory passive time constant:', tau/b.ms, 'ms')
assert abs(tau/b.ms - 20) < 1e-12
# Your turn: compute the inhibitory value using 200 pF and 20 nS.
```
