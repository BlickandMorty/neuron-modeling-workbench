## Read the visual grammar

- A **row** in the activity array is one time point.
- A **column** is one modeled region.
- A region's three coordinates choose where its dot appears.
- The structural matrix chooses which dots receive faint connecting lines.
- Color encodes the model variable at that time. It does not turn that variable into voltage.
- Apparent motion means the regional states changed over time; it is not a movie of charge physically traveling.
