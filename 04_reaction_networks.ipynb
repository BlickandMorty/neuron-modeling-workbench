## Read the saved experiment

```python
runs = []
for manifest in (root / 'results' / 'multiscale').glob('*/manifest.json'):
    item = json.loads(manifest.read_text(encoding='utf-8'))
    if item['lesson'] == 'reaction_network' and item['status'] == 'passed':
        runs.append(manifest.parent)
assert runs, 'No completed example found; follow the reproduction command in the guide.'
saved = sorted(runs)[-1]
print(saved)
display(Image(filename=str(saved / 'reaction_network.png')))
display(json.loads((saved / 'metrics.json').read_text(encoding='utf-8')))
display(json.loads((saved / 'checks.json').read_text(encoding='utf-8')))
```
